This notebook sets initial opinions, random seeds, and figure style. Give the trajectory of three states and distributions at T=50 

```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
import numpy as np

start_dir = Path.cwd().resolve()
for folder in (start_dir, *start_dir.parents):
    if (folder / 'src' / 'monte_carlo.py').is_file():
        repo_root = folder
        sys.path.insert(0, str(repo_root))
        from src.monte_carlo import (
            BiasInitialOpinions, FigureStyle, make_complete_figure,
            make_distribution_figure, make_trajectory_figure,
            print_summary, run_three_states, save_figure,
        )
        break
    if (folder / 'monte_carlo.py').is_file():
        repo_root = folder
        sys.path.insert(0, str(repo_root))
        from monte_carlo import (
            BiasInitialOpinions, FigureStyle, make_complete_figure,
            make_distribution_figure, make_trajectory_figure,
            print_summary, run_three_states, save_figure,
        )
        break
else:
    raise FileNotFoundError('Keep this notebook with monte_carlo.py, or run it inside the repository.')

figure_dir = repo_root / 'figures'
```

## 1. Set the initial opinions and seeds

Each bias group contains 200 non-stubborn agents. In panel (c), the 100 agents with the most negative biases start at $-0.5$; all other non-stubborn agents start at $+0.5$.

```python
group_size = 200

initial_opinions = {
    'bp': BiasInitialOpinions(
        positive=np.full(group_size, 0.5),
        negative=np.full(group_size, -0.5),
    ),
    'bu': BiasInitialOpinions(
        positive=np.full(group_size, 0.5),
        negative=np.full(group_size, 0.5),
    ),
    'ab': BiasInitialOpinions(
        positive=np.full(group_size, 0.5),
        negative=np.r_[np.full(100, -0.5), np.full(100, 0.5)],
    ),
}

seeds = {'bp': 80, 'bu': 80, 'ab': 62}
```

## 2. Run the model

This cell calls the Python model and performs a fresh MC run. It does not load saved final opinions.

```python
results = run_three_states(initial_opinions, seeds, alpha2=0.03)
print_summary(results)
```

## 3. Set the figure style

```python
style = FigureStyle(
    font_family='STIXGeneral',
    kde_bandwidth=0.075,
    histogram_bins=30,
    complete_size=(7.20, 4.70),
)
```

## 4. Trajectories

```python
fig = make_trajectory_figure(results, style)
save_figure(fig, figure_dir / 'three_states_K40_T50_trajectories_from_py')
plt.show()
```

## 5. Terminal distributions

```python
fig = make_distribution_figure(results, style)
save_figure(fig, figure_dir / 'three_states_K40_T50_distributions_from_py')
plt.show()
```
