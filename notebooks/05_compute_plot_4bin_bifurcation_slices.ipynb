# Four-bin mean-field bifurcation slices

Both panels are recomputed directly from the four-bin mean-field equations. No stored branch data are read.


```python
from pathlib import Path
import sys

import matplotlib.pyplot as plt
from IPython.display import Image, display
import numpy as np

root = Path.cwd().resolve()
if root.name == 'notebooks':
    root = root.parent
if not (root / 'src' / 'mean_field_4bin.py').is_file():
    raise RuntimeError('Run this notebook from the repository root or notebooks folder.')
sys.path.insert(0, str(root / 'src'))

import mean_field_4bin as mf
figure_dir = root / 'figures' / 'four_bin'
figure_dir.mkdir(parents=True, exist_ok=True)
```

## Panel (a): vary $\alpha_2$ at $f_s=0.5$


```python
alpha_values = np.arange(0.0, 0.3001, 0.001)
_, alpha_branches, alpha_events = mf.compute_slice('alpha2', alpha_values, 0.50)
print(f'Saddle-node: alpha2 = {alpha_events[0]:.10f}')
print(f'Pitchfork:   alpha2 = {alpha_events[1]:.10f}')
```

```python
fig, ax = plt.subplots(figsize=(4.95, 4.65), dpi=180)
mf.draw_alpha2_panel(ax, alpha_branches, alpha_events)
fig.subplots_adjust(left=0.18, right=0.96, bottom=0.15, top=0.76)
png = figure_dir / 'panel_a_alpha2_bifurcation.png'
pdf = figure_dir / 'panel_a_alpha2_bifurcation.pdf'
fig.savefig(png, dpi=600)
fig.savefig(pdf)
plt.close(fig)
display(Image(filename=str(png)))
```

## Panel (b): vary $f_s$ at $\alpha_2=0$


```python
fs_values = np.arange(0.15, 0.5501, 0.005)
_, fs_branches, fs_events = mf.compute_slice('f_s', fs_values, 0.0)
print(f'Pitchfork:   fs = {fs_events[0]:.10f}')
print(f'Saddle-node: fs = {fs_events[1]:.10f}')
```

```python
fig, ax = plt.subplots(figsize=(4.95, 4.65), dpi=180)
mf.draw_fs_panel(ax, fs_branches, fs_events)
fig.subplots_adjust(left=0.18, right=0.96, bottom=0.15, top=0.76)
png = figure_dir / 'panel_b_fs_bifurcation.png'
pdf = figure_dir / 'panel_b_fs_bifurcation.pdf'
fig.savefig(png, dpi=600)
fig.savefig(pdf)
plt.close(fig)
display(Image(filename=str(png)))
```

## Combined figure


```python
paths = mf.save_results(alpha_branches, alpha_events, fs_branches, fs_events)
display(Image(filename=str(paths['png'])))
for path in paths.values():
    print('Saved:', path)
```
