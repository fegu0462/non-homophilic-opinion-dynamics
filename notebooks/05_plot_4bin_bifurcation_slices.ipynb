# Four-bin mean-field bifurcation diagrams

This notebook recomputes both one-parameter bifurcation diagrams from the four-bin mean-field equations. It does not read precomputed branch or bifurcation data.

- Panel (a): vary $\alpha_2$ at $f_s=0.5$.
- Panel (b): vary $f_s$ at $\alpha_2=0$.

The rendering uses the same fonts, font sizes, state-dependent background fills, stable/unstable branch styles, bifurcation guide lines, legends, and panel geometry as the manuscript figure.

## 1. Setup

```python
from pathlib import Path
import sys
from IPython.display import Image, display


def find_repository_root(start: Path) -> Path:
    start = start.resolve()
    for candidate in (start, *start.parents):
        direct = candidate / 'src' / 'full_reproduction' / 'compute_plot_4bin_two_slices.py'
        nested = candidate / 'paper-code-release' / 'src' / 'full_reproduction' / 'compute_plot_4bin_two_slices.py'
        if direct.exists():
            return candidate
        if nested.exists():
            return candidate / 'paper-code-release'
    raise FileNotFoundError('Could not locate the paper-code-release folder.')


REPO_ROOT = find_repository_root(Path.cwd())
sys.path.insert(0, str(REPO_ROOT))

from src.full_reproduction import compute_plot_4bin_two_slices as reproduce

print('Repository:', REPO_ROOT)
print('Data output:', reproduce.DATA_DIR)
print('Figure output:', reproduce.FIGURE_DIR)
```

## 2. Recompute panel (a): branches versus $\alpha_2$

This cell performs the supplied bidirectional continuation scan, stability calculation, normal-form diagnostics, and automatic bifurcation detection at $f_s=0.5$.

```python
alpha_namespace, alpha_branches, alpha_saddle, alpha_pitchfork = reproduce.compute_alpha_panel()

print(f'Saddle-node: alpha2 = {alpha_saddle:.10f}')
print(f'Pitchfork : alpha2 = {alpha_pitchfork:.10f}')
print('Branch data:', reproduce.ALPHA_CSV)
print('Event data :', reproduce.ALPHA_EVENTS_CSV)
```

### Display panel (a)

```python
alpha_paths = reproduce.render_alpha_panel(
    alpha_branches, alpha_saddle, alpha_pitchfork
)
display(Image(filename=str(alpha_paths['png'])))
for path in alpha_paths.values():
    print('Saved:', path)
```

## 3. Recompute panel (b): branches versus $f_s$

This cell scans the four-bin fixed points at $\alpha_2=0$, tracks the complete branches, calculates their stability, and refines both critical parameter values by solving the fixed-point and singular-Jacobian equations.

```python
fs_records, fs_branches, fs_saddle, fs_pitchfork = reproduce.compute_fs_panel()

print(f'Pitchfork : fs = {fs_pitchfork:.10f}')
print(f'Saddle-node: fs = {fs_saddle:.10f}')
print('Branch data:', reproduce.FS_CSV)
print('Event data :', reproduce.FS_EVENTS_CSV)
```

### Display panel (b)

```python
fs_paths = reproduce.render_fs_panel(fs_branches, fs_saddle, fs_pitchfork)
display(Image(filename=str(fs_paths['png'])))
for path in fs_paths.values():
    print('Saved:', path)
```

## 4. Assemble the two vertically arranged panels

```python
combined_paths = reproduce.render_combined(
    alpha_branches,
    alpha_saddle,
    alpha_pitchfork,
    fs_branches,
    fs_saddle,
    fs_pitchfork,
)

display(Image(filename=str(combined_paths['png'])))
for path in combined_paths.values():
    print('Saved:', path)
```
